## 測試一：原始資料

```python
# HW09 測試一：原始資料

# 1. 建立活動名稱、原始名單與費用
activity_names = ("桌遊體驗", "電影欣賞")

board_raw = " Amy, Ben, amy, Cora "
movie_raw = " ben, Dora, ERIC, ben "

unit_fees = {"桌遊體驗": 120, "電影欣賞": 80}

# 2. 從 tuple 取得活動名稱
board_name = activity_names[0]
movie_name = activity_names[1]

# 3. 清理原始文字，轉成 list
# 移除半形空白、轉成小寫，再用逗號分割
board_list = board_raw.replace(" ", "").lower().split(",")
movie_list = movie_raw.replace(" ", "").lower().split(",")

# 4. 使用 set 刪除重複報名
board_set = set(board_list)
movie_set = set(movie_list)

# 5. 使用集合運算比較兩場活動
both = board_set & movie_set
all_people = board_set | movie_set
board_only = board_set - movie_set
movie_only = movie_set - board_set

# 6. 將名單依照字母順序排序
board_sorted = sorted(board_set)
movie_sorted = sorted(movie_set)
both_sorted = sorted(both)
all_sorted = sorted(all_people)
board_only_sorted = sorted(board_only)
movie_only_sorted = sorted(movie_only)

# 7. 使用切片取得通知名單前兩位
notice_list = all_sorted[:2]

# 8. 使用 dict 保存統計摘要
summary = {
    "桌遊人數": len(board_set),
    "電影人數": len(movie_set),
    "共同人數": len(both),
    "總不重複人數": len(all_people),
    "報名人次": len(board_set) + len(movie_set),
    "桌遊應收": len(board_set) * unit_fees[board_name],
    "電影應收": len(movie_set) * unit_fees[movie_name]
}

# 9. 計算總應收
summary["總應收"] = summary["桌遊應收"] + summary["電影應收"]

# 10. 輸出完整活動報名摘要
print("HW09 活動報名摘要")
print(f"活動：{board_name}，每人 {unit_fees[board_name]:.2f} 元")
print(f"活動：{movie_name}，每人 {unit_fees[movie_name]:.2f} 元")

print(f"桌遊整理串列：{board_list}")
print(f"電影整理串列：{movie_list}")

print(f"桌遊去重名單：{board_sorted}")
print(f"電影去重名單：{movie_sorted}")

print(f"桌遊人數：{summary['桌遊人數']} 人")
print(f"電影人數：{summary['電影人數']} 人")

print(f"共同參加者：{both_sorted}")
print(f"共同人數：{summary['共同人數']} 人")

print(f"只參加桌遊者：{board_only_sorted}")
print(f"只參加電影者：{movie_only_sorted}")

print(f"全部參加者：{all_sorted}")
print(f"總不重複人數：{summary['總不重複人數']} 人")
print(f"報名人次：{summary['報名人次']} 人次")
print(f"通知名單前兩位：{notice_list}")

print(f"桌遊應收：{summary['桌遊應收']:.2f} 元")
print(f"電影應收：{summary['電影應收']:.2f} 元")
print(f"總應收：{summary['總應收']:.2f} 元")
```
